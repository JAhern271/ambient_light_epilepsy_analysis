# 05 - Cleaning LUX data

**Status: not yet written.** Placeholder for wear-time cleaning of the LUX recordings.

Notebook 09 detects non-wear periods in the PAXMIN activity data and extracts valid wear
blocks. The same treatment has not been applied to the LUX data, even though PAX and LUX
come from the same device and therefore share the same non-wear periods.

Light metrics are currently computed over the **whole recording, including non-wear
time**, which will bias any participant with substantial off-wrist periods. Resolving
this is tracked as Phase 4 in `doc/good-practice-plan.md`.
