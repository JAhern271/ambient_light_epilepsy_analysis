# 05 - Cleaning LUX data

**Status: not written, and superseded.**

This was a placeholder for wear-time cleaning of the PAXLUX recordings. PAXLUX-derived
light metrics are computed over the **whole recording, including non-wear time**, which
biases any participant with substantial off-wrist periods.

That gap is now moot for published results. PAXMIN carries minute-level light
(`PAXLXMM`) alongside activity, and notebook 09 already masks **both** to missing over
non-wear minutes, using `PAXTSM < 45` or `PAXPREDM == 3`. So the PAXMIN route resolves
this by construction, and light and activity are cleaned identically because they come
from the same rows.

This notebook would only be worth writing if the PAXLUX analysis were revived. See
`doc/data-sources.md` for how the two light sources differ.
